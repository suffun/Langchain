```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)
```

```python

embedding_function = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
```

```python


doc1 = Document(
    page_content="Delhi is the capital of India.",
    metadata={"source": "india.txt", "topic": "Geography", "page": 1}
)

doc2 = Document(
    page_content="Mumbai is the financial capital of India.",
    metadata={"source": "india.txt", "topic": "Geography", "page": 2}
)

doc3 = Document(
    page_content="Virat Kohli is an Indian cricketer.",
    metadata={"source": "cricket.txt", "topic": "Cricket", "page": 1}
)

doc4 = Document(
    page_content="Python is a popular programming language.",
    metadata={"source": "programming.txt", "topic": "Programming", "page": 1}
)

doc5 = Document(
    page_content="Mango is a tropical fruit commonly eaten in summer.",
    metadata={"source": "fruits.txt", "topic": "Food", "page": 1}
)
```

```python
docs = [doc1,doc2,doc3,doc4,doc5]
```

```python
vector_store = Chroma(
    embedding_function=embedding_function,
    persist_directory="./chroma_db",
    collection_name="my_collection"
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
vector_store.similarity_search(
    query='which fruit is eaten in summer',
    k = 2
)
```

```python
vector_store.similarity_search_with_score(
    query='which fruit is eaten in summer',
    k = 2
)
```

```python
# Metadata filtering
results = vector_store.similarity_search_with_score(
    query="Tell me about fruits",
    filter={"topic": "Food"},
    k=2
)

for doc, score in results:
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("Distance:", score)
    print("-------------------")
```

```python

updated_docs = Document(
    page_content="Delhi is the capital of India And it also most polluted city in inia",
    metadata={"source": "india.txt", "topic": "Geography", "page": 1}
)

vector_store.add_documents('')
```
